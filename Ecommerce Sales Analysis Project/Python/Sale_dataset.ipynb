```python
import pandas as pd
import numpy as np

df = pd.read_csv('sales_data_cleaned.csv')
```

```python
df.describe()
```

```python
df.info()
```

```python
df['order_date'] = pd.to_datetime(df['order_date'], format='mixed', dayfirst=True)
```

```python
df['profit_margin'] = (df['profit'] / df['revenue']) * 100
```

```python
df['order_month'] = df['order_date'].dt.to_period('M').astype(str)
```

```python
df['order_year'] = df['order_date'].dt.year
```

```python
df['delivery_performance'] = pd.cut(
    df['delivery_days'],
    bins=[-1, 2, 5, float('inf')],
    labels=['Fast', 'Normal', 'Delayed']
)
```

```python
df['discount_category'] = pd.cut(
    df['discount_pct'],
    bins = [-1,10,25,50,float('inf')],
    labels = ['Low','Medium','High','Ver High']
)
```

```python
df.to_csv('sales_data_cleaned1.csv', index=False)
```
